```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
import math
```

### Leitura da imagem pelo OpenCV

```python
imagem = cv2.imread("Images/Lenna.png", cv2.IMREAD_GRAYSCALE)
linhas = imagem.shape[0]
colunas = imagem.shape[1] 
print("linhas: ", linhas)
print("colunas: ", colunas)
```

### Exibição da imagem original

```python

# cmap = color maps define mapas de cor para imagem (util para pseudo-cores)
plt.imshow(imagem, cmap='gray', vmin=0, vmax=255)
```

```python
# Profundidade de bits da imagem
# Saída: uint8 (8 bits por canal)
print(imagem.dtype) 
print(imagem.shape)
print(imagem.min(), imagem.max())


print(imagem[0,0])
print(int(imagem[0,0]) * 32 / 256)
```

```python
bits = 1
niveis = 2**bits

imagem_requantizada = np.zeros((linhas, colunas), dtype=np.uint8)

for i in range(linhas):
    for j in range(colunas):
        imagem_requantizada[i,j] = (int(imagem[i, j]) * niveis) / 256

plt.imshow(imagem_requantizada, cmap='gray', vmin=0, vmax=(niveis-1))
```
