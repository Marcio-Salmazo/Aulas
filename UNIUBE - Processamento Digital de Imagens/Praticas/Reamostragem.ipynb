```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
import math
```

### Leitura da imagem pelo OpenCV

```python
imagem = cv2.imread("Images/Lenna.png", cv2.IMREAD_GRAYSCALE)
linhas = imagem.shape[0]
colunas = imagem.shape[1] 
print("linhas: ", linhas)
print("colunas: ", colunas)
```

### Exibição da imagem original

```python
plt.imshow(imagem, cmap='gray', vmin=0, vmax=255)
```

```python
# redução de 96 dpi para 30 dpi - Proporção: 3.2

proporcao = 1
imagem_reamostrada = np.zeros((round(linhas/proporcao), round(colunas/proporcao)))

# Passar os pixel para essa nova amostragem
for i in range(round(linhas/proporcao)):
    for j in range(round(colunas/proporcao)):
        imagem_reamostrada[i][j] = imagem[round(i*proporcao)][round(j*proporcao)] 

# Exibir imagem reamostrada
plt.imshow(imagem_reamostrada, cmap='gray', vmin=0, vmax=255)
```
