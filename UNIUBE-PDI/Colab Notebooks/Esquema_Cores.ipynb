```python
# Bibliotecas
import cv2
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
# Imagem artificial

imagem = np.array([
    [0,   0,   0,   0,   0],
    [0, 255, 255, 113,   0],
    [0, 255,   0, 255,   0],
    [0, 255, 255, 255,   0],
    [0,   0,   0,   0,   0]
], dtype=np.uint8)

plt.imshow(imagem, cmap='gray', vmin=0, vmax=255)
plt.colorbar()
plt.show()
```

```python
altura = 300
largura = 300

vermelho = np.zeros((300, 300, 3), dtype=np.uint8)
verde = np.zeros((300, 300, 3), dtype=np.uint8)
azul = np.zeros((300, 300, 3), dtype=np.uint8)

vermelho[:, :, 2] = 255
verde[:, :, 1] = 255
azul[:, :, 0] = 255

# No esquema BGR do openCV, 0(blue), 1(green), 2(red)
amarelo = vermelho + verde
magenta = vermelho + azul
ciano = verde + azul
branco = vermelho + verde + azul

# Exibição dos resultados:
_, ax = plt.subplots(1, 7)

ax[0].imshow(vermelho)
ax[1].imshow(verde)
ax[2].imshow(azul)
ax[3].imshow(amarelo)
ax[4].imshow(magenta)
ax[5].imshow(ciano)
ax[6].imshow(branco)

for ax in ax.flat:
    ax.set_xticks([])
    ax.set_yticks([])

plt.show()
```

```python
# Carregamento da imagem
imagem = cv2.imread("Lenna.png")
Imagem = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)
R, G, B = cv2.split(imagem)

_, axs = plt.subplots(1, 3)

axs[0].imshow(R, cmap='Reds')
axs[0].set_title('Canal R')
axs[0].axis('off')

axs[1].imshow(G, cmap='Greens')
axs[1].set_title('Canal G')
axs[1].axis('off')

axs[2].imshow(B, cmap='Blues')
axs[2].set_title('Canal B')
axs[2].axis('off')

plt.show()
```

```python
# Retirada de canais RGB

sem_vermelho = imagem.copy()
sem_vermelho[:, :, 0] = 0

sem_verde = imagem.copy()
sem_verde[:, :, 1] = 0

sem_azul = imagem.copy()
sem_azul[:, :, 2] = 0

_, axs = plt.subplots(1, 3)

axs[0].imshow(sem_vermelho)
axs[0].set_title('B+G')
axs[0].axis('off')

axs[1].imshow(sem_verde)
axs[1].set_title('R+B')
axs[1].axis('off')

axs[2].imshow(sem_azul)
axs[2].set_title('R+G')
axs[2].axis('off')
```

```python
hsv = cv2.cvtColor(imagem, cv2.COLOR_BGR2HSV)
H, S, V = cv2.split(hsv)


hsv_colorido = np.zeros_like(hsv)

hsv_colorido[:, :, 0] = H
hsv_colorido[:, :, 1] = 255
hsv_colorido[:, :, 2] = 255
matiz_colorida = cv2.cvtColor(hsv_colorido,cv2.COLOR_HSV2RGB)


_, axs = plt.subplots(1, 3)

axs[0].imshow(matiz_colorida)
axs[0].set_title('matiz')
axs[0].axis('off')

axs[1].imshow(S, cmap='gray')
axs[1].set_title('saturação')
axs[1].axis('off')

axs[2].imshow(V, cmap='gray')
axs[2].set_title('valor')
axs[2].axis('off')

plt.show()
```
