```python
# Bibliotecas
import cv2
import numpy as np
import math
import matplotlib.pyplot as plt

# Carregamento da imagem em escala cinza
imagem = cv2.imread("Lenna.png", cv2.IMREAD_GRAYSCALE)

linhas = imagem.shape[0]
colunas = imagem.shape[1]
print("linhas: ", linhas)
print("colunas: ", colunas)
```

```python
proporcao = 4
imagem_reamostrada = np.zeros((round(linhas/proporcao),
                               round(colunas/proporcao)))

# Passar os pixel selecionados para essa nova amostragem
for i in range(round(linhas/proporcao)):
    for j in range(round(colunas/proporcao)):
        imagem_reamostrada[i,j] = imagem[round(i*proporcao), round(j*proporcao)]
```

```python
# Exibição lado-a-lado

# subplots(1, 2) --> 1 linha e 2 colunas
# _ significa 'qualquer coisa'
_, ax = plt.subplots(1, 2)

ax[0].imshow(imagem, cmap='gray', vmin=0, vmax=255)
ax[0].set_title("Imagem Original")

ax[1].imshow(imagem_reamostrada, cmap='gray', vmin=0, vmax=255)
ax[1].set_title("Imagem Reamostrada")

plt.show()
```
