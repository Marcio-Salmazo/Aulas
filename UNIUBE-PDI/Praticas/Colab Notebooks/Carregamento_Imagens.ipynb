```python
# Demais bibliotecas
import cv2
import matplotlib.pyplot as plt

# Carregamento e exibição da Imagem
path = '/content/drive/MyDrive/Colab Notebooks/Images/Lenna.png'
image = cv2.imread(path)

print("linhas: ", image.shape[0])
print("colunas: ", image.shape[1])

# O openCV usa o formato BGR
formatted = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(formatted, cmap='gray', vmin=0, vmax=255)
```
