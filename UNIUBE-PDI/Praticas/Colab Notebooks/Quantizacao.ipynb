```python
# Bibliotecas
import cv2
import numpy as np
import math
import matplotlib.pyplot as plt

# Carregamento da imagem em escala cinza
imagem = cv2.imread("Lenna.png", cv2.IMREAD_GRAYSCALE)

linhas = imagem.shape[0]
colunas = imagem.shape[1]
print("linhas: ", linhas)
print("colunas: ", colunas)
```

```python
profundiade = 1
niveis = 2**profundiade
imagem_requantizada = np.zeros((linhas, colunas), dtype=np.uint8)

# Passar os pixel selecionados para essa nova amostragem
for i in range(linhas):
    for j in range(colunas):
        imagem_requantizada[i,j] = (int(imagem[i,j]) * niveis)/256
```

```python
# Exibição lado-a-lado
_, ax = plt.subplots(1, 2)

ax[0].imshow(imagem, cmap='gray', vmin=0, vmax=255)
ax[0].set_title("Imagem Original")

ax[1].imshow(imagem_requantizada, cmap='gray', vmin=0, vmax=(niveis-1))
ax[1].set_title("Imagem Reamostrada")

plt.show()
```
